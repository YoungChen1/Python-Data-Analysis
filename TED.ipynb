# 评估和清理TED数据

## 分析目标

此数据分析的目的是，根据TED数据，挖掘最受欢迎的演讲主题，以便分析当今社会人群的关注热点。

本实战项目的目的在于练习评估数据干净和整洁度，并且基于评估结果，对数据进行清洗，从而得到可供下一步分析的数据。

## 简介

该数据集是国际思想传播平台 TED（Ideas Worth Spreading）演讲的结构化汇总数据，系统记录了多场 TED 演讲的基础信息、演讲者背景、传播表现及受众反馈，是研究全球思想传播趋势、公众兴趣偏好及 TED 平台生态的优质样本。

1. 演讲基础信息
- `title`：演讲标题，直观概括演讲核心议题；
- `name`：演讲的完整名称（含演讲者与主题），便于快速定位演讲身份。
- `description`：演讲内容简介，详细说明演讲者观点或演讲主旨；
- `duration`：演讲时长（单位：秒）；
- `event`：演讲所属 TED 活动，标记演讲的举办场景与年份；
- `film_date/published_date`：均为 Unix 时间戳（秒级），分别对应演讲拍摄日期与平台发布日期；
2. 演讲者信息
- `main_speaker`：主要演讲者姓名，涵盖科学家、社会活动家、艺术家、企业家等多元身份；
- `speaker_occupation`：演讲者职业，反映演讲者的专业背景；
- `num_speaker`：单场演讲的演讲者人数（默认值为 1，少数为多人合作演讲）。
3. 传播与互动表现
- `views`：演讲观看次数，范围从数万到数千万，直接体现演讲的传播影响力；
- `comments`：演讲评论数量，反映受众对演讲的讨论热度；
- `languages`：演讲提供的翻译语言种类，体现演讲的全球传播广度。
4. 受众反馈与主题分类
- `ratings`：受众评价列表，包含 14 类评价维度及对应评价次数，可用于分析受众偏好；
- `tags`：演讲主题标签，涵盖教育、科技、环境、心理学、文化等领域，是主题挖掘的核心字段；
- `related_talks`：相关演讲列表，记录与当前演讲主题关联的其他演讲信息（如演讲 ID、标题、观看量），体现内容间的关联性；
- `url`：演讲在 TED 官网的直接链接，支持跳转查看原始演讲视频或页面。

## 清理数据

根据前面数据评估结论，我们需要进行的数据清理工作包括：

- 对`ratings`变量进行行拆分
- 对`related_talks`变量进行行拆分
- 对`speaker_occupation`变量进行行拆分
- 对`tags`变量进行行拆分
- 将`film_date`变量数据类型转换为日期时间
- 将`published_date`变量数据类型转换为日期时间

为了区分开经过清理后的数据和原始数据，我们创建新的变量`cleaned_ted_data`，让它为`ted_data`复制出的副本，我们之后的清理操作都将被运用到`cleaned_ted_data`上。

```python
cleaned_ted_data = ted_data.copy()
cleaned_ted_data.head()
```

对`ratings`、`related_talks`、`tags`变量进行行拆分：

`ratings`、`related_talks`、`tags`三个变量表面上看是列表，实际数据类型为字符串，需使用Python内置的`eval`函数，把字符串转换成表达式，可以帮我们把表示列表的字符串转换成列表本身。

```python
cleaned_ted_data["ratings"] = cleaned_ted_data["ratings"].apply(lambda val: eval(val))
```

```python
cleaned_ted_data["related_talks"] = cleaned_ted_data["related_talks"].apply(lambda val: eval(val))
```

```python
cleaned_ted_data["tags"] = cleaned_ted_data["tags"].apply(lambda val: eval(val))
```

```python
cleaned_ted_data = cleaned_ted_data.explode(["ratings"])
```

```python
cleaned_ted_data = cleaned_ted_data.explode(["related_talks"])
```

```python
cleaned_ted_data = cleaned_ted_data.explode(["tags"])
```

```python
cleaned_ted_data.info()
```

对`speaker_occupation`变量进行行拆分：

检查原始数据集中`speaker_occupation`不同职业出现的次数。

```python
ted_data["speaker_occupation"].value_counts()
```

可见，`speaker_occupation`共有1458种，通过随机抽取10行，观察`speaker_occupation`变量每个单元格内不同职业的分隔方式。

```python
ted_data["speaker_occupation"].sample(10)
```

经多组观察，每个单元格内不同职业可以","、"/"、"and"、"+"进行分隔。分隔后对每个职业的首字母大写。

```python
import re
# 定义处理函数：拆分职业并统一首字母大写
def process_occupation(occupation):
    if pd.isna(occupation):
        return []

    # 正则表达式增加对+的处理，现在包含：,、/、and、+四种分隔符
    split_occupations = re.split(r'[,/+]\s*|\s+and\s+', str(occupation))

    # 处理每个职业：去除空格并首字母大写
    processed = []
    for occ in split_occupations:
        occ = occ.strip()  # 去除前后空格
        if occ:  # 确保不是空字符串
            occ = occ.capitalize()  # 首字母大写，其余小写
            processed.append(occ)

    return processed
```

```python
cleaned_ted_data["speaker_occupation"] = cleaned_ted_data["speaker_occupation"].apply(process_occupation)
```

```python
cleaned_ted_data = cleaned_ted_data.explode("speaker_occupation")
```

```python
cleaned_ted_data["speaker_occupation"].sample(10)
```

经多次观察，符合预期。

将`film_date`变量数据类型转换为日期时间

```python
cleaned_ted_data["film_date"] = pd.to_datetime(cleaned_ted_data["film_date"])
cleaned_ted_data["film_date"]
```

将`published_date`变量数据类型转换为日期时间

```python
cleaned_ted_data["published_date"] = pd.to_datetime(cleaned_ted_data["published_date"])
cleaned_ted_data["published_date"]
```
